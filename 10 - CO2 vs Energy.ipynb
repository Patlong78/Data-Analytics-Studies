```python
## Block 52: Complete Emission Analysis (CO2 per kW)

import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Ensure we have the data
if 'co2_per_kw' not in df.columns:
    df['co2_per_kw'] = pd.to_numeric(df['CO2 Emissions (Mg)'], errors='coerce') / \
                       pd.to_numeric(df['Total Plant Installed Capacity (MW)'], errors='coerce')
    df = df.dropna(subset=['co2_per_kw'])

grouped = df.groupby('Fuel Types')['co2_per_kw']

# Calculate all metrics
p10 = grouped.quantile(0.10)
p90 = grouped.quantile(0.90)
trimmed_mean = grouped.apply(lambda x: x[(x >= x.quantile(0.10)) & (x <= x.quantile(0.90))].mean())
mean = grouped.mean()
std_dev = grouped.std()
variance = grouped.var()
cv = (std_dev / mean) * 100  # Coefficient of variation
spread_percent = ((p90 - p10) / trimmed_mean) * 100

# Create summary DataFrame
summary = pd.DataFrame({
    'Trimmed Mean (10-90%)': trimmed_mean,
    'Regular Mean': mean,
    'Std Dev': std_dev,
    'Variance': variance,
    'CV %': cv,
    'P10': p10,
    'P90': p90,
    'Spread %': spread_percent
}).round(4)

print("=" * 60)
print("EMISSION ANALYSIS SUMMARY: CO2 per kW by Fuel Type")
print("=" * 60)
print(summary)
print("\n")

# Visualization 1: Bar chart of Spread %
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# Plot 1: Spread % comparison
ax1 = axes[0, 0]
colors = ['#2d2d2d', '#8B4513', '#4169E1']
spread_percent.plot(kind='bar', ax=ax1, color=colors)
ax1.set_title("Emission Spread (10-90% Range) by Fuel Type", fontsize=12, fontweight='bold')
ax1.set_ylabel("Spread %")
ax1.set_xlabel("Fuel Type")
ax1.tick_params(axis='x', rotation=0)
for i, v in enumerate(spread_percent):
    ax1.text(i, v + 1, f"{v:.1f}%", ha='center', fontsize=10)

# Plot 2: Trimmed Mean comparison
ax2 = axes[0, 1]
trimmed_mean.plot(kind='bar', ax=ax2, color=colors)
ax2.set_title("Average CO2/kW (10-90% Trimmed)", fontsize=12, fontweight='bold')
ax2.set_ylabel("kg CO2/kW")
ax2.set_xlabel("Fuel Type")
ax2.tick_params(axis='x', rotation=0)
for i, v in enumerate(trimmed_mean):
    ax2.text(i, v + 0.5, f"{v:.2f}", ha='center', fontsize=10)

# Plot 3: Coefficient of Variation
ax3 = axes[1, 0]
cv.plot(kind='bar', ax=ax3, color=colors)
ax3.set_title("Coefficient of Variation (CV %)", fontsize=12, fontweight='bold')
ax3.set_ylabel("CV %")
ax3.set_xlabel("Fuel Type")
ax3.tick_params(axis='x', rotation=0)
for i, v in enumerate(cv):
    ax3.text(i, v + 1, f"{v:.1f}%", ha='center', fontsize=10)

# Plot 4: Box plot of distribution
ax4 = axes[1, 1]
df.boxplot(column='co2_per_kw', by='Fuel Types', ax=ax4)
ax4.set_title("CO2/kW Distribution by Fuel Type", fontsize=12, fontweight='bold')
ax4.set_ylabel("kg CO2/kW")
ax4.set_xlabel("Fuel Type")
plt.suptitle('')  # Remove automatic title

plt.tight_layout()
plt.show()

# Print detailed findings
print("=" * 60)
print("KEY FINDINGS:")
print("=" * 60)
for fuel in trimmed_mean.index:
    print(f"\n{fuel}:")
    print(f"  • Typical Range (10–90%): {p10[fuel]:.4f} → {p90[fuel]:.4f} kg CO2/kW")
    print(f"  • Spread: {spread_percent[fuel]:.1f}%")
    print(f"  • Coefficient of Variation: {cv[fuel]:.1f}%")
    print(f"  • Higher CV = more complex data = harder to classify")
```

```python
ults# Define a specificity calculation function that will be used for metrics
from sklearn.metrics import confusion_matrix
import numpy as np

def compute_specificity(y_true, y_pred, average='weighted'):
    """Compute specificity (true negative rate) for multi-class problems."""
    labels = np.unique(np.concatenate([y_true, y_pred]))
    cm = confusion_matrix(y_true, y_pred, labels=labels)
    # True negatives: total - (false positives + true positives + false negatives)
    tn = cm.sum() - cm.sum(axis=0) - cm.sum(axis=1) + np.diag(cm)
    fp = cm.sum(axis=0) - np.diag(cm)
    # Avoid division by zero
    class_specificity = np.divide(tn, tn + fp, out=np.zeros_like(tn, dtype=float), where=(tn + fp) != 0)
    
    if average == 'weighted':
        support = cm.sum(axis=1)
        return np.sum(class_specificity * support) / support.sum() if support.sum() != 0 else 0.0
    elif average == 'macro':
        return np.mean(class_specificity)
    return class_specificity
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Select metrics to compare
metrics_performance = ["Accuracy", "Precision", "Sensitivity", "F1 Score"]
metrics_all = ["Accuracy", "Specificity", "Precision", "Sensitivity", "F1 Score", "Time (sec)"]
models = df_results["Model"].tolist()

# Create figure with subplots for performance metrics (2x3 grid)
fig, axes = plt.subplots(2, 3, figsize=(16, 10))
fig.suptitle("Model Performance Comparison (All Metrics)", fontsize=16, fontweight='bold')
axes = axes.flatten()

# Plot each metric
for idx, metric in enumerate(metrics_all):
    ax = axes[idx]
    values = df_results[metric].values
    colors = plt.cm.viridis(np.linspace(0, 1, len(models)))
    
    bars = ax.barh(models, values, color=colors)
    
    # Set x-axis limits based on metric type
    if metric == "Time (sec)":
        ax.set_xlabel("Time (seconds)", fontsize=11)
        ax.set_xlim(0, max(values) * 1.15)
    else:
        ax.set_xlabel(metric, fontsize=11)
        ax.set_xlim(0, 1)
    
    ax.set_title(metric, fontsize=12, fontweight='bold')
    
    # Add value labels
    for i, (bar, value) in enumerate(zip(bars, values)):
        if metric == "Time (sec)":
            label_text = f'{value:.4f}s'
        else:
            label_text = f'{value:.3f}'
        ax.text(value + (max(values) * 0.02), i, label_text, va='center', fontsize=9)

plt.tight_layout()
plt.show()

# Create grouped bar chart with all performance metrics including Time and Precision
metrics_grouped = ["Accuracy", "Precision", "Specificity", "Sensitivity", "F1 Score", "Time (sec)"]
fig, ax = plt.subplots(figsize=(16, 6))

x = np.arange(len(models))
width = 0.13
multiplier = 0

for idx, metric in enumerate(metrics_grouped):
    offset = width * multiplier
    # Scale time to be visible on same chart (divide by 10 for better visualization)
    if metric == "Time (sec)":
        values = df_results[metric].values / 10  # Scale down time for visibility
        ax.bar(x + offset, values, width, label=f'{metric} (÷10)', alpha=0.8)
    else:
        ax.bar(x + offset, df_results[metric].values, width, label=metric, alpha=0.8)
    multiplier += 1

ax.set_xlabel("Model", fontsize=12, fontweight='bold')
ax.set_ylabel("Score", fontsize=12, fontweight='bold')
ax.set_title("Model Performance Metrics Comparison (All Metrics Including Time & Precision)", fontsize=14, fontweight='bold')
ax.set_xticks(x + width * 2.5)
ax.set_xticklabels(models, rotation=45, ha='right')
ax.legend(loc='upper right', fontsize=9, ncol=2)
ax.set_ylim(0, 1.1)
ax.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()

# Create separate bar chart for training time
fig, ax = plt.subplots(figsize=(10, 6))

colors_time = plt.cm.cool(np.linspace(0, 1, len(models)))
bars = ax.barh(models, df_results["Time (sec)"], color=colors_time)

ax.set_xlabel("Training Time (seconds)", fontsize=12, fontweight='bold')
ax.set_title("Model Training Time Comparison", fontsize=14, fontweight='bold')

# Add value labels
for i, (bar, value) in enumerate(zip(bars, df_results["Time (sec)"])):
    ax.text(value + (df_results["Time (sec)"].max() * 0.02), i, f'{value:.4f}s', va='center', fontsize=10)

plt.tight_layout()
plt.show()

# Print summary table
print("\n" + "="*80)
print("MODEL PERFORMANCE SUMMARY")
print("="*80)
print(df_results.to_string(index=False))
```
